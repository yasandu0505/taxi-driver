```python
import gymnasium as gym
import numpy as np

env = gym.make("Taxi-v4", render_mode="ansi")
state, info = env.reset(seed=0)

print("States:", env.observation_space.n)  
print("Actions:", env.action_space.n) 
print(env.render())

# env.unwrapped.encode(4, 3, 2, 0)

# encoding the state (taxi_row, taxi_col, passenger_location, destination)
print(env.unwrapped.encode(4, 3, 2, 0))

# decoding the state
print(list(env.unwrapped.decode(328)))
```
